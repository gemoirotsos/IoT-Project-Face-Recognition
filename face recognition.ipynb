```python
import face_recognition

# 1. Encode known people from training images
known_face_encodings = []
known_face_names = []

training_images = {
    "me1.jpg": "Giorgos",
    "me2.jpg": "Giorgos",
    "moir1.jpg": "Moirotsos",
    "moir2.jpg": "Moirotsos"
}

for filename, name in training_images.items():
    image = face_recognition.load_image_file(filename)
    encodings = face_recognition.face_encodings(image)
    if encodings:
        known_face_encodings.append(encodings[0])
        known_face_names.append(name)

# 2. Load and encode the test image
test_image = face_recognition.load_image_file("test.jpg")
test_encodings = face_recognition.face_encodings(test_image)

if not test_encodings:
    print("❌ No face found in the test image.")
else:
    test_encoding = test_encodings[0]

    # 3. Compare the test face with known encodings
    results = face_recognition.compare_faces(known_face_encodings, test_encoding)
    distances = face_recognition.face_distance(known_face_encodings, test_encoding)

    if True in results:
        best_match_index = distances.argmin()
        print(f"✅ Match found: {known_face_names[best_match_index]} (distance={distances[best_match_index]:.4f})")
    else:
        print("❌ No match found.")
```
